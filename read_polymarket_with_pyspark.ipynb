# Read Polymarket Sports-Betting Events (Parquet) with PySpark

This notebook reads the parquet generated by `polymarket_sports_events.py` and validates the schema.

```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName('PolymarketSportsRead').getOrCreate()
```

```python
parquet_path = 'data/polymarket_sports_events.parquet'
df = spark.read.parquet(parquet_path)
df.printSchema()
df.show(10, truncate=False)
```

```python
print(f'Row count: {df.count()}')
spark.stop()
```
